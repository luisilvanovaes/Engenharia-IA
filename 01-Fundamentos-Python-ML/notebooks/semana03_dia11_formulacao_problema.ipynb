# Semana 03 — Dia 11: Formulação de Problemas de Machine Learning

## Projeto: AI Energy Auditor

### Objetivo
Definir o problema de previsão do consumo energético de uma fábrica virtual, identificando variável-alvo, variáveis explicativas, baseline e métricas de avaliação.

### Contexto
Piloto industrial com dados simulados, visando futura validação em empresas reais.

## Exercício 61 — Formulação do problema

### 1. Problema de negócio

Uma fábrica precisa prever seu consumo diário de energia elétrica
para melhorar o planejamento operacional, controlar custos e
identificar possíveis desperdícios.

### 2. Objetivo do modelo

Prever o consumo de energia elétrica (kWh) do próximo dia,
utilizando dados históricos e informações operacionais
disponíveis no momento da previsão.

### 3. Formulação de Machine Learning

- Aprendizado: supervisionado
- Problema: regressão
- Target: consumo diário de energia (kWh)
- Features: consumo histórico, volume de produção, turnos
  e outras variáveis operacionais conhecidas antecipadamente
- Horizonte: próximo dia
- Baseline: último consumo diário conhecido
- Métricas: MAE e RMSE

### 4. Aplicação

Projeto AI Energy Auditor — piloto industrial com dados simulados.

### 5. Definição das variáveis

#### Variável-alvo (Target)

consumo_kwh: consumo total de energia elétrica
da fábrica durante o próximo dia.

#### Variáveis explicativas (Features)

- consumo_kwh_anterior: consumo registrado no dia anterior.
- media_consumo_7d: média dos últimos sete dias.
- producao_prevista: volume de produção planejado.
- turnos_previstos: quantidade de turnos programados.
- temperatura_prevista: temperatura ambiente prevista.
- dia_semana: dia da semana da previsão.

#### Cuidados com os dados

As variáveis explicativas devem estar disponíveis
antes do período que desejamos prever.

Não utilizaremos o consumo real do dia previsto
como entrada do modelo, pois isso causaria
vazamento de dados (Data Leakage).


### 6. Definição do Baseline

#### Método: Persistência

O baseline será um modelo simples que utiliza
o consumo energético do dia anterior como
previsão para o próximo dia.

Fórmula:

Consumo previsto (t+1) = Consumo real (t)

Exemplo:

- Consumo de hoje: 12.500 kWh
- Previsão para amanhã: 12.500 kWh

#### Objetivo

Estabelecer uma referência de desempenho para
comparar os modelos de Machine Learning.

Um modelo mais complexo somente será considerado
vantajoso se apresentar melhorias relevantes em
relação ao baseline, avaliadas com dados não
utilizados no treinamento.

#### Métricas de avaliação

- MAE: erro absoluto médio, expresso em kWh.
- RMSE: raiz do erro quadrático médio, em kWh.

Quanto menores os valores, melhor o desempenho
preditivo, considerando o mesmo conjunto de avaliação.


### 7. Análise dos custos dos erros

#### Subestimação do consumo

Ocorre quando o consumo real é maior que o previsto.

Possíveis consequências:
- Planejamento energético insuficiente.
- Desvios no orçamento operacional.
- Dificuldade para antecipar períodos de maior demanda.

#### Superestimação do consumo

Ocorre quando o consumo real é menor que o previsto.

Possíveis consequências:
- Planejamento excessivamente conservador.
- Projeções financeiras acima do necessário.
- Decisões inadequadas sobre recursos energéticos.

#### Avaliação dos erros

Erro = Consumo real - Consumo previsto

- Erro positivo: subestimação.
- Erro negativo: superestimação.
- Erro zero: previsão exata.

#### Critério de negócio

Além de MAE e RMSE, avaliaremos separadamente
os erros de subestimação e superestimação.

Em uma futura implantação industrial, os custos
financeiros serão definidos com base nas tarifas,
contratos e condições reais da empresa.


### 8. Estratégia de validação temporal

#### Separação cronológica dos dados

Os dados serão organizados em ordem temporal,
sem embaralhamento aleatório.

Divisão inicial proposta:

- Treinamento: primeiros 70% dos registros.
- Validação: 15% dos registros seguintes.
- Teste: últimos 15% dos registros.

Os percentuais poderão ser ajustados conforme
o tamanho e a frequência do conjunto de dados.

#### Prevenção de Data Leakage

- Não utilizar informações futuras no treinamento.
- Calcular variáveis históricas somente com dados passados.
- Ajustar transformações usando apenas os dados permitidos
  em cada etapa de treinamento.
- Preservar o conjunto de teste para a avaliação final.

#### Walk-Forward Validation

Posteriormente, utilizaremos validação walk-forward,
treinando com dados anteriores e avaliando previsões
em períodos posteriores.

#### Critério de comparação

Todos os modelos serão comparados utilizando
as mesmas janelas temporais e métricas,
incluindo o baseline de persistência.



### 9. Critérios de sucesso do modelo

#### Critérios técnicos

1. Apresentar MAE e RMSE inferiores aos do baseline
   no mesmo conjunto de avaliação.

2. Manter desempenho consistente durante a
   validação temporal.

3. Não utilizar informações futuras nas previsões.

4. Avaliar separadamente os erros de subestimação
   e superestimação.

5. Registrar parâmetros, métricas e resultados
   para garantir a reprodutibilidade.

#### Critérios de negócio

1. Fornecer previsões úteis ao planejamento energético.

2. Identificar desvios relevantes entre consumo
   previsto e consumo realizado.

3. Permitir estimativas de impacto financeiro
   com base em tarifas e contratos definidos.

4. Produzir resultados compreensíveis e auditáveis.

#### Limitações do piloto

Os dados utilizados inicialmente serão simulados.

Resultados positivos nesse ambiente não comprovam
desempenho equivalente em uma fábrica real.

A implantação comercial exigirá validação com
dados reais, avaliação operacional e requisitos
adicionais de segurança e confiabilidade.

#### Decisão

O modelo somente será candidato à implantação
quando atender aos critérios técnicos e de negócio
estabelecidos para o ambiente de aplicação.
