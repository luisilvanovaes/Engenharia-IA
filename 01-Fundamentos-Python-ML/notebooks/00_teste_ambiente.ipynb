```python

import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

print("=== AMBIENTE DE ENGENHARIA DE IA ===")
print("Python:", sys.version.split()[0])
print("Executável:", sys.executable)
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)

# Criar um pequeno conjunto de dados
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([3, 5, 7, 9, 11])

# Treinar um modelo de regressão linear
modelo = LinearRegression()
modelo.fit(X, y)

# Realizar uma previsão
previsao = modelo.predict([[6]])

print("\n=== TESTE MACHINE LEARNING ===")
print(f"Coeficiente: {modelo.coef_[0]:.2f}")
print(f"Intercepto: {modelo.intercept_:.2f}")
print(f"Previsão para X=6: {previsao[0]:.2f}")

# Gerar gráfico
plt.figure(figsize=(8, 4))
plt.scatter(X.ravel(), y, label="Dados reais")
plt.plot(X.ravel(), modelo.predict(X),
         label="Regressão linear")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Primeiro modelo de Machine Learning")
plt.legend()
plt.grid(True)
plt.show()

print("\nTESTE CONCLUÍDO!")
```
