```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.arange(0, 20, 0.1)
y = np.sin(x)
plt.figure(figsize=(9, 4))
plt.plot(x, y)
plt.title("Primeiro gráfico - Engenharia de IA")
plt.xlabel("X")
plt.ylabel("sen(X)")
plt.grid(True)
plt.show()
print("Jupyter configurado com sucesso!")
```
